```python
from pyspark.sql.functions import *
from pyspark.sql import functions as F
```

```python
catalogo = 'lakehouse_vertice'
medalhao = 'bronze'
tabela = 'pmo'

df_pmo = spark.read.table(f'{catalogo}.{medalhao}.{tabela}')
```

```python
df_pmo_silver = df_pmo.withColumn(
    'lucro_previsto',
    round((col('valor_contrato') - col('orcamento_custo')),2)
)
df_pmo_silver = df_pmo_silver.withColumn(
        'margem_pct',
    (round((col('lucro_previsto') / col('valor_contrato')), 2) * 100)
)
```

```python
catalogo = 'lakehouse_vertice'
medalhao = 'silver'
tabela = 'pmo'

df_pmo_silver.write \
    .mode('overwrite') \
    .saveAsTable(f'{catalogo}.{medalhao}.{tabela}')

print(f'Tabela {catalogo}.{medalhao}.{tabela} criada com sucesso')
```
