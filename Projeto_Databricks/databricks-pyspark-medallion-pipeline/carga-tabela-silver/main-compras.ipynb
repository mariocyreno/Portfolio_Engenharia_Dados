```python
from pyspark.sql.functions import *
from pyspark.sql import functions as F
```

```python
catalogo = 'lakehouse_vertice'
medalhao = 'bronze'
tabela = 'compras'

df_compras = spark.read.table(f'{catalogo}.{medalhao}.{tabela}')
```

```python
df_compras = df_compras.withColumn(
    'atraso',
    when(col("data_entrega") > col("data_prometida"), lit('Atrasado'))
    .otherwise(lit("No Prazo"))
)
df_compras = df_compras.withColumn(
    "dias_atraso",
    date_diff(col("data_entrega"), col("data_prometida"))
         )
df_compras = df_compras.withColumn(
    "dias_atraso",
    when(col("dias_atraso") < 0, lit(0))
    .otherwise(col("dias_atraso"))
)
```

```python
df_compras_silver = df_compras

catalogo = 'lakehouse_vertice'
medalhao = 'silver'
tabela = 'compras'

df_compras_silver.write \
    .mode('overwrite') \
    .saveAsTable(f'{catalogo}.{medalhao}.{tabela}'
    )

print(f'Tabela {catalogo}.{medalhao}.{tabela} criada com sucesso!')
```
