```python
from pyspark.sql.functions import *
from pyspark.sql import functions as F
```

```python
catalogo = 'lakehouse_vertice'
medalhao = 'bronze'
tabela = 'colaboradores'

df_colaboradores = spark.read.table(f'{catalogo}.{medalhao}.{tabela}')
```

```python
df_colaboradores = df_colaboradores.withColumn(
    'ativo',
    when(col("data_desligamento").isNull(), 'Ativo')
    .otherwise(lit('Desligado'))
)
```

```python
df_colaboradores_silver = df_colaboradores[
    'colaborador_id',
    'nome',
    'papel',
    'senioridade',
    'custo_hora',
    'data_admissao',
    'data_desligamento',
    'ativo'
] 
```

```python
catalogo = 'lakehouse_vertice'
medalhao = 'silver'
tabela = 'colaboradores'

df_colaboradores_silver.write \
    .mode('overwrite') \
    .saveAsTable(f'{catalogo}.{medalhao}.{tabela}'
    )

print(f'Tabela {catalogo}.{medalhao}.{tabela} criada com sucesso!')
```
