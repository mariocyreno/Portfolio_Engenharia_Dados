```python
import numpy as np
from sklearn import datasets
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.model_selection import KFold
```

```python
#Carregando os dados e separando as entradas e saidas
wine = datasets.load_wine()
x = wine.data
y = wine.target

print('# Linha: ', x.shape[0])
print('# Coluna: ', x.shape[1])
print('# Classe: ', len(np.unique(y)))
```

```python
#Separar entrada e saida em treinamento e teste
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.4, random_state=42)
print('Treinamento (60%): ',len(x_train))
print('Teste (40%): ', len(x_test))

#Realizando o aprendizado do modelo testado
clf = LogisticRegression(solver='lbfgs', max_iter=10000)
clf.fit(x_train, y_train)

#Classificando os dados não-vistos no modelo de teste
y_pred = clf.predict(x_test)

#Comparar com resultados reais e preditos
print('Acuracia %.2f' % accuracy_score(y_test, y_pred))
```

```python
acc = []

#Realizar a aprendizagem do modelo com K-Fold Cross Validation
kd = KFold(n_splits=5)

for train_index, test_index in kd.split(x):
    x_train, y_train = x[train_index], y[train_index]
    x_test, y_test = x[test_index], y[test_index]

#realizando o aprendizado do modelo testado
clf = LogisticRegression(solver='lbfgs', max_iter=10000)
clf.fit(x_train, y_train)

#classificando os dados não-vistos no modelo de teste
y_pred = clf.predict(x_test)

#Calcula e armazena a acuracia do modelo
acc.append(accuracy_score(y_test, y_pred))
```

```python
print('Acuracia %.2f' % np.mean(acc), "Desvio padrão:", np.std(acc))
```
